## 1. Functions that call other functions

````python
def make_prompt(topic):
    return f"Write one fact about {topic}"

def fake_llm(prompt):
    # pretend LLM: it just answers in capitals
    return f"ANSWER TO: {prompt.upper()}"

def run(topic):
    prompt = make_prompt(topic)     # step 1
    answer = fake_llm(prompt)       # step 2 uses the result of step 1
    return answer

print(run("black holes"))
````

## 2. Functions are values too
You can store a function in a variable or pass it to another function (without the `()`).

````python
def shout(text):
    return text.upper() + "!"

def whisper(text):
    return text.lower() + "..."

def apply(func, text):
    return func(text)              # call whatever function we were given

print(apply(shout, "hello"))
print(apply(whisper, "HELLO"))
````

## 3. `lambda` — a tiny function in one line
`lambda x: x * 2` is the same as `def double(x): return x * 2`.

````python
double = lambda x: x * 2
add_exclamation = lambda text: text + "!"

print(double(5))
print(add_exclamation("Chains are fun"))
````

## 4. A pipeline = a list of functions run in order

````python
steps = [
    lambda topic: f"Tell me about {topic}",   # 1. build the prompt
    fake_llm,                                  # 2. call the "model"
    lambda text: text.strip().title(),         # 3. "parse" / clean the output
]

value = "cricket"
for step in steps:
    value = step(value)            # the output of one step becomes the input of the next
    print("->", value)
````

## 5. `if / elif / else` — making decisions
`4_conditional_chain.py` chooses a different reply depending on the sentiment. That is just an `if`.

````python
def reply_to(feedback):
    sentiment = "positive" if "love" in feedback else "negative"   # one-line if
    if sentiment == "positive":
        return "Thank you so much! 😊"
    elif "late" in feedback:
        return "Sorry for the delay — we are fixing it."
    else:
        return "Sorry to hear that. How can we help?"

print(reply_to("I love this phone"))
print(reply_to("Delivery was late"))
print(reply_to("It broke"))
````

## 6. Parallel steps — two jobs on the same input
`3_parallel_chain.py` makes *notes* and a *quiz* from the same text, then merges them. We can do the same with a dictionary.

````python
text = "Python is a popular language. It is easy to read. It is used in AI."

branches = {
    "notes": lambda t: t.split(". ")[0],          # first sentence
    "word_count": lambda t: len(t.split()),       # number of words
}

results = {}
for name, func in branches.items():   # .items() gives (key, value) pairs
    results[name] = func(text)

print(results)
````

## 7. 🔗 Build a mini `|` chain
In LangChain you write `prompt | model | parser`. The `|` works because the classes define a special method called `__or__`. Here is a tiny version:

````python
class Step:
    def __init__(self, func):
        self.func = func

    def __or__(self, other):                      # runs when Python sees  a | b
        return Step(lambda x: other.func(self.func(x)))

    def invoke(self, x):                          # same name LangChain uses!
        return self.func(x)

prompt = Step(lambda topic: f"Explain {topic} simply")
model  = Step(fake_llm)
parser = Step(lambda text: text.lower())

chain = prompt | model | parser
print(chain.invoke("neural networks"))
````
