## 1. `dataclass`
`@dataclass` writes `__init__` and a nice print for you. Perfect for simple data.

````python
from dataclasses import dataclass, field

@dataclass
class Section:
    title: str
    description: str
    content: str = ""            # default value

s = Section("Introduction", "What LangGraph is")
print(s)
print(s.title)
````

## 2. A plan = a list of sections
In `01_planning_agent_deep_research_report.ipynb` the LLM returns a plan like this (as a Pydantic model).

````python
plan = [
    Section("Introduction", "What state management means"),
    Section("Checkpointers", "How LangGraph saves progress"),
    Section("Conclusion", "Key takeaways"),
]

for number, section in enumerate(plan, start=1):
    print(f"{number}. {section.title} — {section.description}")
````

## 3. Multi-line f-strings
Triple quotes let a string span many lines — great for prompts and reports.

````python
def write_section(section):
    # a real agent asks the LLM to write this text
    return f'''## {section.title}

This section explains **{section.description.lower()}**.
'''

plan[0].content = write_section(plan[0])
print(plan[0].content)
````

## 4. Joining sections into one Markdown report

````python
from datetime import date

for section in plan:
    section.content = write_section(section)

report = f"# LangGraph Report\n_Generated on {date.today()}_\n\n" + "\n".join(s.content for s in plan)
print(report)
````

## 5. Saving the report to a file

````python
from pathlib import Path

out = Path("demo_report.md")
out.write_text(report, encoding="utf-8")
print(out.exists(), "-", len(out.read_text(encoding="utf-8")), "characters saved")
out.unlink()             # delete the demo file (unlink = delete)
````

## 6. 🔗 Combining results from parallel writers
Each worker writes **one** section at the same time, so they can finish in any order. We put them back in order with `sorted`.

````python
import operator

finished = []                                     # like Annotated[list, operator.add]
finished = operator.add(finished, [("Conclusion", 3)])
finished = operator.add(finished, [("Introduction", 1)])
finished = operator.add(finished, [("Checkpointers", 2)])

for title, order in sorted(finished, key=lambda pair: pair[1]):
    print(order, title)
````
