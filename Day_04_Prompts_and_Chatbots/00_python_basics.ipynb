## 1. Dictionaries
A **dictionary** stores **key → value** pairs inside curly brackets `{ }`.
Think of a real dictionary: the *word* is the key, the *meaning* is the value.

````python
message = {"role": "user", "content": "What is an LLM?"}

print(message["role"])          # get a value by its key
print(message["content"])

message["content"] = "What is RAG?"    # change a value
message["time"] = "10:30"              # add a new key
print(message)

print(message.get("language", "not set"))   # .get() gives a default instead of an error
````

## 2. A list of dictionaries = chat history
Every chat model receives the conversation as a **list of messages**. Each message is a dictionary with a `role` and `content`:
- `system` → instructions for the AI
- `user` → what the human says
- `assistant` → what the AI answered

````python
chat_history = [
    {"role": "system", "content": "You are a helpful teacher."},
    {"role": "user", "content": "Hi!"},
    {"role": "assistant", "content": "Hello! What do you want to learn?"},
]

chat_history.append({"role": "user", "content": "Teach me Python."})   # new message

for msg in chat_history:
    print(f"{msg['role']:>9}: {msg['content']}")
````

## 3. Prompt templates
A template is text with **blanks** `{like_this}`. `.format()` fills them in.
LangChain's `PromptTemplate` does the same thing.

````python
template = "Explain {topic} to a {audience} in {length} sentences."

prompt1 = template.format(topic="transformers", audience="child", length=2)
prompt2 = template.format(topic="embeddings", audience="engineer", length=5)

print(prompt1)
print(prompt2)
````

## 4. `while` loops and `break`
A `while` loop repeats **while** a condition is true. `break` stops it.
A chatbot is just a loop: *read message → reply → repeat until the user types `exit`*.

> In the real `7_chatbot.py`, the user types with `input()`. Here we use a list of pretend messages so the notebook runs by itself.

````python
pretend_user_messages = ["hello", "what is AI?", "exit", "this is never read"]

history = []
i = 0
while True:                         # loop forever...
    user_text = pretend_user_messages[i]
    i = i + 1
    if user_text == "exit":         # ...until the user says exit
        print("Bot: Goodbye!")
        break                       # leave the loop
    reply = f"You said '{user_text}'"   # a real bot would call the LLM here
    history.append({"role": "user", "content": user_text})
    history.append({"role": "assistant", "content": reply})
    print("Bot:", reply)

print("Messages remembered:", len(history))
````

## 5. Saving and loading JSON files
**JSON** is a text format that looks like Python dictionaries. It is used everywhere for saving data and talking to APIs.
`5_prompt_generator.py` saves a template to `template.json` exactly like this.

> 💡 **Always add `encoding="utf-8"`** when you open a text file. Windows uses a different default, so emojis or Hindi text would crash with `UnicodeDecodeError` without it.

````python
import json

data = {"template": "Summarise {text}", "input_variables": ["text"]}

# write to a file ("w" = write mode)
with open("demo_template.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2)

# read it back ("r" = read mode)
with open("demo_template.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)

print(loaded["template"])
print(loaded["input_variables"])
````

## 6. Reading a text file line by line

````python
# Create a small text file first
with open("demo_history.txt", "w", encoding="utf-8") as f:
    f.write("user: hi\n")
    f.write("assistant: hello!\n")

with open("demo_history.txt", "r", encoding="utf-8") as f:
    for line in f:                  # one line at a time
        print(line.strip())         # .strip() removes the newline at the end
````
