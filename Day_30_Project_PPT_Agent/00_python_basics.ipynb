## 1. A blueprint = nested data
A list of slides; each slide is a dictionary; `bullets` is a list inside it.

````python
blueprint = {
    "title": "Intro to AI Agents",
    "slides": [
        {"layout": "title", "title": "Intro to AI Agents", "subtitle": "Day 30 demo"},
        {"layout": "bullets", "title": "What is an agent?",
         "bullets": ["An LLM that can use tools", "It plans, acts and observes", "It loops until done"]},
        {"layout": "bullets", "title": "Popular frameworks",
         "bullets": ["LangGraph", "Google ADK", "Deep Agents"]},
    ],
}
print(blueprint["slides"][1]["bullets"][0])      # dict -> list -> dict -> list -> item
````

## 2. Looping through nested data

````python
for number, slide in enumerate(blueprint["slides"], start=1):
    print(f"Slide {number}: {slide['title']}")
    for bullet in slide.get("bullets", []):      # .get(..., []) -> the title slide has no bullets
        print("    •", bullet)
````

## 3. Pretty-printing JSON
The agent saves this to `generated_Blueprint_Plans/latest_presentation_blueprint.json`.

````python
import json
print(json.dumps(blueprint, indent=2)[:300], "...")
````

## 4. 🔗 Create a real PowerPoint

````python
from pptx import Presentation
from pptx.util import Pt

prs = Presentation()
TITLE_LAYOUT, BULLET_LAYOUT = prs.slide_layouts[0], prs.slide_layouts[1]

for spec in blueprint["slides"]:
    if spec["layout"] == "title":
        slide = prs.slides.add_slide(TITLE_LAYOUT)
        slide.shapes.title.text = spec["title"]
        slide.placeholders[1].text = spec["subtitle"]
    else:
        slide = prs.slides.add_slide(BULLET_LAYOUT)
        slide.shapes.title.text = spec["title"]
        body = slide.placeholders[1].text_frame
        body.text = spec["bullets"][0]                    # first bullet
        for bullet in spec["bullets"][1:]:                # the rest
            p = body.add_paragraph()
            p.text = bullet
            p.font.size = Pt(24)

prs.save("demo_agents.pptx")
print("Saved demo_agents.pptx ✅ (open it to have a look before the next cell deletes it)")
````

## 5. Read it back

````python
check = Presentation("demo_agents.pptx")
print("Number of slides:", len(check.slides))
for s in check.slides:
    print("-", s.shapes.title.text)

import os
os.remove("demo_agents.pptx")
````
