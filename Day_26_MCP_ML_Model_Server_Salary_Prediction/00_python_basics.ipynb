## 1. A pandas DataFrame
A **DataFrame** is a table (like Excel) inside Python. `train_model.py` reads `sample_salary_dataset.csv` into one.

````python
import pandas as pd

df = pd.DataFrame({
    "experience": [1, 2, 3, 5, 7, 10, 12],
    "education":  ["Bachelor", "Bachelor", "Master", "Master", "Bachelor", "PhD", "Master"],
    "salary":     [40000, 45000, 60000, 72000, 80000, 120000, 115000],
})
df
````

````python
print(df.shape)          # (rows, columns)
print(df.columns.tolist())
df.head(3)               # the first 3 rows
````

## 2. Selecting and filtering

````python
print(df["salary"].mean())                       # average of one column
print(df["salary"].max())

seniors = df[df["experience"] >= 5]                # keep rows where the condition is True
seniors
````

## 3. Grouping

````python
df.groupby("education")["salary"].mean().round()
````

## 4. Training a model
**Linear regression** draws the best straight line through the points: *more experience → higher salary*.

````python
from sklearn.linear_model import LinearRegression

X = df[["experience"]]        # input(s) — double brackets = a table with one column
y = df["salary"]              # what we want to predict

model = LinearRegression()
model.fit(X, y)               # learn from the data

new = pd.DataFrame({"experience": [4, 8]})
print(model.predict(new).round())
print(f"Each extra year adds about {model.coef_[0]:.0f} to the salary")
````

## 5. Saving and loading with `pickle`
`salary_model.pkl` in this folder is a saved model. `pickle` turns a Python object into a file and back.

````python
import pickle, os

with open("demo_model.pkl", "wb") as f:      # "wb" = write bytes
    pickle.dump(model, f)

with open("demo_model.pkl", "rb") as f:      # "rb" = read bytes
    loaded = pickle.load(f)

print(loaded.predict(pd.DataFrame({"experience": [6]})).round())
os.remove("demo_model.pkl")
````

## 6. 🔗 Wrap the model in a tool
In `salary_mcp_server.py` a function like this gets `@mcp.tool` on top, so an AI agent can call it. The type hints and docstring tell the agent how to use it.

````python
def predict_salary(experience_years: float) -> dict:
    """Predict the yearly salary for a given number of years of experience."""
    if experience_years < 0:
        return {"error": "experience cannot be negative"}
    value = loaded.predict(pd.DataFrame({"experience": [experience_years]}))[0]
    return {"experience_years": experience_years, "predicted_salary": round(float(value))}

print(predict_salary(3))
print(predict_salary(-1))
````
