## 1. Importing modules
A **module** is a file full of ready-made code. `import` lets you use it.

````python
import math                 # built-in maths tools
import random               # built-in random-number tools

print(math.sqrt(16))        # square root -> 4.0
print(random.randint(1, 6)) # a random dice roll between 1 and 6
````

## 2. Writing your own function
A **function** is a named block of code you can reuse.
```
def function_name(input1, input2):
    ...do something...
    return result
```

````python
def greet(name):
    # 'name' is the input (called a parameter)
    message = f"Hello, {name}!"
    return message          # 'return' sends the result back

result = greet("Ali")
print(result)
print(greet("Mira"))
````

## 3. Default values
A parameter can have a **default value**. If you don't pass it, the default is used.
This is exactly how `ChatOpenAI(model="gpt-4o-mini", temperature=0.7)` works.

````python
def make_model(name, temperature=0.7):
    return f"Model '{name}' with temperature {temperature}"

print(make_model("gpt-4o-mini"))                 # uses the default 0.7
print(make_model("gpt-4o-mini", temperature=0))  # we choose 0
````

## 4. Formatting decimal numbers
`{value:.2f}` shows a number with 2 digits after the dot.

````python
score = 0.876543
print(f"Similarity: {score}")
print(f"Similarity: {score:.2f}")    # 0.88
print(f"Similarity: {score:.0%}")    # 88%
````

## 6. Randomness and temperature
An LLM picks the next word using **probabilities**.
- Low temperature → almost always picks the most likely word (boring but safe).
- High temperature → more surprising choices (creative).

````python
import random

words = ["blue", "cloudy", "purple"]
low_temp  = [0.90, 0.09, 0.01]   # 'blue' is very likely
high_temp = [0.40, 0.35, 0.25]   # choices are more even

random.seed(42)   # makes the random results repeatable
print("Low temperature :", random.choices(words, weights=low_temp, k=8))
print("High temperature:", random.choices(words, weights=high_temp, k=8))
````

## 7. 🔗 Connect to today: embeddings
An **embedding** turns text into a list of numbers. Similar meanings → similar numbers.
We measure similarity with **cosine similarity** (1.0 = same direction, 0 = unrelated).

````python
import math

def dot(a, b):
    # multiply matching numbers and add them up
    total = 0
    for x, y in zip(a, b):      # zip pairs the lists together
        total = total + x * y
    return total

def length(a):
    return math.sqrt(dot(a, a))

def cosine_similarity(a, b):
    return dot(a, b) / (length(a) * length(b))

# Tiny made-up embeddings (real ones have 384-3072 numbers!)
cat    = [0.9, 0.8, 0.1]
kitten = [0.85, 0.75, 0.2]
car    = [0.1, 0.2, 0.95]

print(f"cat vs kitten: {cosine_similarity(cat, kitten):.2f}")   # high
print(f"cat vs car   : {cosine_similarity(cat, car):.2f}")      # low
````
