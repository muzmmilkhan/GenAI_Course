## 1. Cloud options as data
(Example prices only — real prices change, always check the provider's website.)

````python
options = {
    "AWS Lambda":      {"per_hour": 0.0,  "per_million_requests": 0.20, "gpu": False, "always_on": False},
    "AWS App Runner":  {"per_hour": 0.064, "per_million_requests": 0.0, "gpu": False, "always_on": True},
    "AWS ECS Fargate": {"per_hour": 0.05,  "per_million_requests": 0.0, "gpu": False, "always_on": True},
    "AWS EC2 g5":      {"per_hour": 1.006, "per_million_requests": 0.0, "gpu": True,  "always_on": True},
    "GCP Cloud Run":   {"per_hour": 0.0,  "per_million_requests": 0.40, "gpu": False, "always_on": False},
}
print(len(options), "options")
````

## 2. A cost function

````python
HOURS_PER_MONTH = 730

def monthly_cost(option, requests_per_month):
    hours = HOURS_PER_MONTH if option["always_on"] else 0
    return option["per_hour"] * hours + option["per_million_requests"] * requests_per_month / 1_000_000

print(round(monthly_cost(options["AWS Lambda"], 2_000_000), 2))
print(round(monthly_cost(options["AWS EC2 g5"], 2_000_000), 2))
````

## 3. Sort and choose

````python
traffic = 3_000_000           # requests per month

costs = {name: monthly_cost(o, traffic) for name, o in options.items()}
ranked = sorted(costs.items(), key=lambda pair: pair[1])

cheapest_name, cheapest_cost = ranked[0]
print(f"Cheapest: {cheapest_name} at ${cheapest_cost:.2f}/month")
````

## 4. Filter by requirements

````python
def recommend(needs_gpu, traffic):
    # if a GPU is needed keep only GPU options, otherwise keep everything
    candidates = {n: o for n, o in options.items() if o["gpu"] or not needs_gpu}
    return min(candidates, key=lambda n: monthly_cost(candidates[n], traffic))

print("Small chatbot API      ->", recommend(needs_gpu=False, traffic=500_000))
print("Self-hosted fine-tuned ->", recommend(needs_gpu=True, traffic=500_000))
````

## 5. A neat comparison table

````python
print(f"{'Service':<17}{'GPU':^6}{'Always on':^11}{'$/month':>10}")
print("-" * 44)
for name, cost in ranked:
    o = options[name]
    print(f"{name:<17}{'yes' if o['gpu'] else '-':^6}{'yes' if o['always_on'] else '-':^11}{cost:>10.2f}")
````
