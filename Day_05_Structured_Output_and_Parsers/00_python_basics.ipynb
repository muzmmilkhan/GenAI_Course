## 1. Type hints
A **type hint** says what kind of value a variable or parameter should hold. Python does not force it, but tools (and LLMs!) read it.

````python
name: str = "Gemini"
version: float = 2.5
tags: list[str] = ["fast", "cheap"]

def add(a: int, b: int) -> int:     # '-> int' means the function returns an int
    return a + b

print(add(2, 3))
````

## 2. `Optional` and `Literal`
- `Optional[str]` → a string **or** `None` (nothing).
- `Literal["pos", "neg"]` → only these exact values are allowed.

````python
from typing import Optional, Literal

def review_label(stars: int) -> Literal["positive", "negative"]:
    if stars >= 3:
        return "positive"
    return "negative"

nickname: Optional[str] = None      # no nickname yet
print(review_label(5), review_label(1), nickname)
````

## 3. `TypedDict` — a dictionary with a fixed shape

````python
from typing import TypedDict

class Review(TypedDict):
    summary: str
    sentiment: str

r: Review = {"summary": "Great phone", "sentiment": "positive"}
print(r["summary"], "->", r["sentiment"])
````

## 4. Classes and objects
A **class** is a blueprint. An **object** is something built from that blueprint.
- `__init__` runs when you create the object.
- `self` means "this object".

````python
class Movie:
    def __init__(self, title, rating):
        self.title = title          # attribute (data stored in the object)
        self.rating = rating

    def describe(self):             # method (a function inside the class)
        return f"{self.title} is rated {self.rating}/10"

m1 = Movie("Inception", 9)
m2 = Movie("Cats", 3)
print(m1.describe())
print(m2.title)
````

## 5. Pydantic models
**Pydantic** is like a class with superpowers: it **checks** the data for you.
`with_structured_output(MyModel)` in today's lesson makes the LLM return one of these.

````python
from pydantic import BaseModel, Field

class Person(BaseModel):
    name: str
    age: int = Field(gt=0, description="Age in years, must be positive")
    city: str = "Unknown"           # default value

p = Person(name="Riya", age="25")   # "25" (text) is converted to 25 automatically
print(p)
print(p.age + 1)
print(p.model_dump())               # turn it into a normal dictionary
````

````python
from pydantic import ValidationError

try:
    bad = Person(name="Tom", age=-5)       # age must be > 0
except ValidationError as error:
    print("❌ Pydantic caught a mistake:")
    print(error.errors()[0]["msg"])
````

## 6. 🔗 Connect to today: JSON text → Python data
An LLM always answers with **text**. If we ask for JSON, we can turn that text into a dictionary with `json.loads()` and then check it with Pydantic.

````python
import json

llm_answer = '{"name": "Arjun", "age": 30, "city": "Pune"}'   # pretend this came from an LLM

data = json.loads(llm_answer)        # text -> dictionary
print(type(data), data["city"])

person = Person(**data)              # ** unpacks the dictionary into keyword arguments
print(person)

print(person.model_dump_json())      # Pydantic object -> JSON text again
````
