## 1. Dataclasses for hotels and flights

````python
from dataclasses import dataclass, field, asdict

@dataclass
class Hotel:
    name: str
    city: str
    price_per_night: float
    rating: float

@dataclass
class Flight:
    airline: str
    origin: str
    destination: str
    price: float

hotels = [
    Hotel("Sea View", "Bali", 80, 4.5),
    Hotel("Budget Inn", "Bali", 35, 3.8),
    Hotel("Royal Palace", "Bali", 250, 4.9),
    Hotel("City Stay", "Tokyo", 120, 4.2),
]
flights = [Flight("AirAsia", "Delhi", "Bali", 320), Flight("Singapore Air", "Delhi", "Bali", 540)]
print(hotels[0])
````

## 2. Filtering and sorting objects
These are the *search tools* the hotel sub-agent uses.

````python
def search_hotels(city, max_price):
    found = [h for h in hotels if h.city == city and h.price_per_night <= max_price]
    return sorted(found, key=lambda h: h.rating, reverse=True)     # best rated first

for h in search_hotels("Bali", max_price=100):
    print(f"{h.name:<12} ${h.price_per_night}/night ⭐ {h.rating}")
````

## 3. A `Trip` class with methods

````python
@dataclass
class Trip:
    traveller: str
    budget: float
    bookings: list = field(default_factory=list)    # a fresh empty list for every trip

    def spent(self):
        return sum(b["cost"] for b in self.bookings)

    def remaining(self):
        return self.budget - self.spent()

    def book(self, item, cost):
        if cost > self.remaining():
            raise ValueError(f"Cannot book {item}: only ${self.remaining()} left")
        self.bookings.append({"item": item, "cost": cost})

trip = Trip("Sachin", budget=1500)
trip.book("Flight AirAsia", flights[0].price)
trip.book("Sea View x 5 nights", hotels[0].price_per_night * 5)
print("Spent:", trip.spent(), "| Left:", trip.remaining())
````

## 4. Over budget → error

````python
try:
    trip.book("Royal Palace x 5 nights", 250 * 5)
except ValueError as e:
    print("⚠️", e)
````

## 5. 🔗 Objects → dictionaries → JSON
FastAPI (`app.py`) sends JSON to the browser. `asdict()` turns a dataclass into a dictionary.

````python
import json

trip_dict = asdict(trip)
print(trip_dict)
print(json.dumps(trip_dict, indent=2))
````
