## 1. Creating a list
A **list** keeps many values in order, inside square brackets `[ ]`.

````python
models = ["BERT", "GPT", "T5"]     # a list of 3 strings
numbers = [10, 20, 30, 40]          # a list of numbers

print(models)
print("How many models?", len(models))
````

## 2. Indexing — get one item
Python counts from **0**!
```
models = ["BERT", "GPT", "T5"]
index:      0       1      2
negative:  -3      -2     -1
```

````python
print(models[0])    # first item  -> BERT
print(models[1])    # second item -> GPT
print(models[-1])   # last item   -> T5 (negative = count from the end)
````

## 3. Slicing — get several items
`my_list[start:stop]` gives items from `start` up to **but not including** `stop`.

````python
words = ["I", "love", "learning", "about", "AI"]

print(words[0:2])   # ['I', 'love']
print(words[2:])    # from index 2 to the end
print(words[:3])    # from the start up to index 3 (not included)
print(words[-2:])   # the last two words
````

## 4. Changing a list
Lists can be changed: add, insert, replace and remove items.

````python
tasks = ["read", "write"]

tasks.append("translate")     # add at the end
tasks.insert(0, "classify")   # insert at position 0
tasks[1] = "summarise"        # replace the item at index 1
tasks.remove("write")         # remove by value

print(tasks)
print("translate" in tasks)   # check if an item exists -> True
````

## 5. Looping over a list
- `for item in my_list:` repeats the indented code for every item.
- `range(n)` gives the numbers `0, 1, ..., n-1`.

````python
for model in models:
    print("Model:", model)

print("---")
for i in range(3):              # i = 0, 1, 2
    print("Index", i, "holds", models[i])
````

## 6. Joining a list back into a string
`" ".join(list)` glues the words together with a space between them.

````python
words = ["Transformers", "are", "powerful"]
sentence = " ".join(words)
print(sentence)
````

## 7. 🔗 Connect to today
**BERT** hides a word with `[MASK]` and learns to guess it (it looks left *and* right).
**GPT** only looks left and guesses the **next** word.

````python
sentence = "the cat sat on the mat"
tokens = sentence.split()

# BERT style: hide the word at index 2
masked = tokens.copy()        # copy so the original list is not changed
masked[2] = "[MASK]"
print("BERT sees: ", " ".join(masked))
print("Answer is: ", tokens[2])
````

````python
# GPT style: show the words one by one and ask "what comes next?"
for i in range(1, len(tokens)):
    seen_so_far = " ".join(tokens[:i])     # everything to the left
    next_word = tokens[i]                  # the word the model must guess
    print(f"{seen_so_far:<20} -> next word: {next_word}")
````
