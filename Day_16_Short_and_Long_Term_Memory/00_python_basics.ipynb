## 1. Keep the last N messages
LLMs have a limited **context window**, so old messages must be removed.

````python
messages = [f"message {i}" for i in range(1, 11)]   # message 1 ... message 10

print(messages[-3:])          # only the last 3
print(messages[:2])           # the first 2 (e.g. the system prompt)
````

## 2. Trim by a word budget
`trim_messages` in LangChain counts **tokens**. We count words: start from the newest message and stop when the budget is full.

````python
chat = [
    "Hi, I am Kabir and I live in Mumbai",
    "Nice to meet you Kabir",
    "I like cricket and coding",
    "Great hobbies!",
    "What city do I live in?",
]

def trim(history, max_words=12):
    kept, used = [], 0
    for msg in reversed(history):         # newest first
        words = len(msg.split())
        if used + words > max_words:
            break
        kept.insert(0, msg)               # put it back in the original order
        used += words
    return kept

print(trim(chat))     # notice: the city information was trimmed away!
````

## 3. Summarise instead of deleting
Better idea: turn the old messages into a short **summary** and keep that.

````python
old, recent = chat[:-2], chat[-2:]
summary = "Summary: " + "; ".join(m.split(",")[0] for m in old)   # a real app asks the LLM to summarise
print([summary] + recent)
````

## 4. Tuples as keys
A **tuple** is like a list that cannot change: `("user_1", "memories")`. Because it cannot change, it can be a dictionary key.
LangGraph's `InMemoryStore` uses tuples like this as **namespaces**.

````python
namespace = ("user_42", "memories")
print(namespace[0], type(namespace))

user_id, kind = namespace          # unpacking a tuple into two variables
print(user_id, kind)
````

## 5. 🔗 Long-term memory with nested dictionaries

````python
store = {}

def put(namespace, key, value):
    store.setdefault(namespace, {})       # create the inner dict if it is missing
    store[namespace][key] = value

put(("user_42", "memories"), "name", "Kabir")
put(("user_42", "memories"), "city", "Mumbai")
put(("user_7", "memories"), "name", "Lina")

print(store[("user_42", "memories")]["city"])   # Mumbai — remembered across sessions
print(store)
````

## 6. Searching and updating memories

````python
def search(namespace, word):
    memories = store.get(namespace, {})
    return {k: v for k, v in memories.items() if word.lower() in f"{k} {v}".lower()}   # dict comprehension

print(search(("user_42", "memories"), "mumbai"))

store[("user_42", "memories")]["city"] = "Pune"     # update (Kabir moved!)
del store[("user_7", "memories")]["name"]           # delete a memory
print(store)
````
