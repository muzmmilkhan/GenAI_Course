## 1. Default and keyword arguments

````python
def book_hotel(city, nights=1, stars=3):
    return f"{nights} night(s) in a {stars}-star hotel in {city}"

print(book_hotel("Goa"))
print(book_hotel("Goa", nights=3))
print(book_hotel(stars=5, city="Paris"))     # with names, order doesn't matter
````

## 2. `*args` — any number of values

````python
def total_cost(*prices):
    print("prices is a tuple:", prices)
    return sum(prices)

print(total_cost(120, 80))
print(total_cost(120, 80, 45, 10))
````

## 3. `**kwargs` — any number of named values

````python
def make_request(**details):
    print("details is a dict:", details)
    return ", ".join(f"{k}={v}" for k, v in details.items())

print(make_request(origin="Delhi", destination="Dubai", passengers=2))
````

## 4. Returning several values
`return a, b` returns a **tuple**, which you can unpack into separate variables.

````python
def cheapest(flights):
    best = min(flights, key=lambda f: f["price"])
    return best["airline"], best["price"]

airline, price = cheapest([{"airline": "IndiGo", "price": 210}, {"airline": "Emirates", "price": 450}])
print(airline, price)
````

## 5. Specialist functions (the sub-agents)

````python
def flight_agent(request):
    return f"✈️ Found a flight for: {request}"

def hotel_agent(request):
    return f"🏨 Found a hotel for: {request}"

def weather_agent(request):
    return f"⛅ Weather report for: {request}"

specialists = {
    "flight": flight_agent,
    "hotel": hotel_agent,
    "weather": weather_agent,
}
````

## 6. 🔗 The supervisor
The supervisor reads the request, decides **which** specialists are needed, calls them and collects their answers.

````python
keywords = {
    "flight": ["fly", "flight", "plane"],
    "hotel": ["hotel", "stay", "room"],
    "weather": ["weather", "rain", "sunny"],
}

def supervisor(request):
    text = request.lower()
    chosen = [name for name, words in keywords.items() if any(w in text for w in words)]
    print("🧑‍💼 Supervisor picks:", chosen)
    results = {}
    for name in chosen:
        results[name] = specialists[name](request)     # delegate
    return results

answers = supervisor("I want to fly to Bali and stay in a hotel. Will it rain?")
for name, answer in answers.items():
    print(f"{name:<8} -> {answer}")
````

## ✅ Summary

| Concept | Example |
|---|---|
| Default argument | `def f(city, nights=1)` |
| Keyword call | `f(city='Goa', nights=3)` |
| Many values | `def f(*args)` |
| Many named values | `def f(**kwargs)` |
| Return many | `return name, price` |
| Unpack | `airline, price = cheapest(...)` |
| Delegate | `specialists[name](request)` |
