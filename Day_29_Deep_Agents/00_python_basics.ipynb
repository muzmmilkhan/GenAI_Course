## 1. Folders and files with `pathlib`

````python
from pathlib import Path

workspace = Path("demo_workspace")
(workspace / "notes").mkdir(parents=True, exist_ok=True)     # parents=True creates the folders on the way

(workspace / "notes" / "todo.md").write_text("- read filings\n- compute ratios\n", encoding="utf-8")
(workspace / "summary.md").write_text("# Summary\n", encoding="utf-8")

for path in sorted(workspace.rglob("*")):                     # rglob = search all sub-folders too
    kind = "📁" if path.is_dir() else "📄"
    print(kind, path)
````

## 2. A *virtual* file system
Deep Agents can keep files **in memory** instead of on disk: a dictionary of `path → content`.

````python
vfs = {}

def write_file(path, content):
    vfs[path] = content
    return f"wrote {len(content)} characters to {path}"

def read_file(path):
    return vfs.get(path, f"ERROR: {path} not found")

def ls():
    return sorted(vfs)

print(write_file("/reports/memo.md", "# Memo"))
print(write_file("/data/notes.txt", "net income looks strong"))
print(ls())
print(read_file("/data/notes.txt"))
print(read_file("/missing.txt"))
````

## 3. A to-do list with statuses
The `write_todos` tool lets the agent plan big tasks and tick them off.

````python
todos = [
    {"task": "Load JPM 2023 statements", "status": "done"},
    {"task": "Compute profitability ratios", "status": "in_progress"},
    {"task": "Check for red flags", "status": "pending"},
    {"task": "Write IC memo", "status": "pending"},
]

def show(todos):
    icons = {"done": "✅", "in_progress": "⏳", "pending": "⬜"}
    for t in todos:
        print(icons[t["status"]], t["task"])

show(todos)
todos[1]["status"] = "done"; todos[2]["status"] = "in_progress"
print("--- after an update ---")
show(todos)
````

## 4. Financial ratios as functions
(Approximate JPMorgan 2023 numbers, in billions of dollars.)

````python
company = {"revenue": 158.1, "net_income": 49.6, "total_assets": 3875.4, "equity": 327.9}

def net_margin(c):
    return c["net_income"] / c["revenue"]

def return_on_equity(c):
    return c["net_income"] / c["equity"]

def return_on_assets(c):
    return c["net_income"] / c["total_assets"]

print(net_margin(company), return_on_equity(company))
````

## 5. Percentages and rounding

````python
ratios = {
    "Net margin": net_margin(company),
    "Return on equity": return_on_equity(company),
    "Return on assets": return_on_assets(company),
}
for name, value in ratios.items():
    print(f"{name:<18} {value:.2%}   (rounded: {round(value, 3)})")
````

## 6. 🔗 Writing an audit memo

````python
flags = [name for name, v in ratios.items() if v < 0.02]      # anything under 2% gets a closer look

memo = "# Investment Committee Memo — JPM 2023\n\n## Key ratios\n"
memo += "\n".join(f"- **{n}:** {v:.2%}" for n, v in ratios.items())
memo += "\n\n## Items to review\n" + ("\n".join(f"- {f}" for f in flags) or "- none")

(workspace / "memo.md").write_text(memo, encoding="utf-8")
print((workspace / "memo.md").read_text(encoding="utf-8"))
````
