## 1. Comments and `print()`
- A **comment** starts with `#`. Python ignores it — it is a note for humans.
- `print()` shows something on the screen.

````python
# This is a comment. Python skips this line.
print("Hello, Generative AI!")   # print shows text on the screen
print(2026)                      # it can print numbers too
````

## 2. Variables
A **variable** is a labelled box that stores a value. You create one with `=`.
```
name_of_box = value
```

````python
model_name = "GPT"        # a box called model_name stores the text "GPT"
layers = 12               # a box called layers stores the number 12

print(model_name)
print(layers)

# You can change what is inside the box
layers = 24
print("Now layers is:", layers)
````

## 3. Numbers and maths
Python has two main number types:
- `int` → whole numbers like `5`, `100`
- `float` → decimal numbers like `0.7`, `3.14`

````python
a = 10
b = 3

print(a + b)    # addition       -> 13
print(a - b)    # subtraction    -> 7
print(a * b)    # multiplication -> 30
print(a / b)    # division       -> 3.333...
print(a // b)   # whole-number division -> 3
print(a % b)    # remainder      -> 1
print(a ** 2)   # power (10 squared) -> 100
````

## 4. Strings (text)
Text in Python is called a **string**. Put it inside quotes: `"like this"` or `'like this'`.

````python
sentence = "Transformers changed AI"

print(sentence.upper())      # ALL CAPITALS
print(sentence.lower())      # all small letters
print(len(sentence))         # how many characters (letters + spaces)
print("AI" in sentence)      # is "AI" inside the sentence? -> True
````

## 5. f-strings — mixing text and variables
Put an `f` before the quotes and write variables inside `{ }`.

````python
model = "BERT"
params_millions = 110

print(f"{model} has about {params_millions} million parameters.")
````

## 6. What type is it?
`type()` tells you what kind of value you have.

````python
print(type(12))          # int    -> whole number
print(type(0.7))         # float  -> decimal number
print(type("hello"))     # str    -> text
print(type(True))        # bool   -> True / False
````

## 7. 🔗 Connect to today: text → tokens
An LLM cannot read a sentence directly. First it **splits text into small pieces called tokens**.
Real tokenizers are smarter, but splitting on spaces gives you the idea.

````python
text = "the cat sat on the mat"

tokens = text.split()          # split the text at every space
print(tokens)                  # this is a list of words (we learn lists tomorrow)
print("Number of tokens:", len(tokens))
````

````python
# Give every word a number (an ID) — models work with numbers, not words!
# A "for loop" repeats the indented lines once for every token.
for token in tokens:
    print(token, "->", len(token), "letters")
````
